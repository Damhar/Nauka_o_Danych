### Zadanie 40. Łączenie ramek danych: merge i concat
Kod poniżej wykonuje operację na danych wariantu 14 (lub na niewielkiej ramce demonstracyjnej, gdy polecenie dotyczy tworzenia ramki).

````python
# Łączenie ramek danych: merge i concat
lewa = df[["risk_id", "risk_name"]].drop_duplicates()
prawa = df.groupby("risk_id", as_index=False)["mean"].mean().rename(columns={"mean": "srednia_mean"})
polaczone_merge = pd.merge(lewa, prawa, on="risk_id", how="inner")
display(polaczone_merge.head())
polaczone_concat = pd.concat([df.head(3), df.tail(3)], ignore_index=True)
display(polaczone_concat)
````

### Zadanie 41. Nowe kolumny przez działania matematyczne
Kod poniżej wykonuje operację na danych wariantu 14 (lub na niewielkiej ramce demonstracyjnej, gdy polecenie dotyczy tworzenia ramki).

````python
# Nowe kolumny przez działania matematyczne
df_matematyka = df.copy()
df_matematyka["szerokosc_przedzialu"] = df_matematyka["upper"] - df_matematyka["lower"]
df_matematyka["srodek_przedzialu"] = (df_matematyka["upper"] + df_matematyka["lower"]) / 2
display(df_matematyka[["lower", "upper", "szerokosc_przedzialu", "srodek_przedzialu"]].head())
````

### Zadanie 42. Nowa kolumna przy użyciu lambda

### Zadanie 43. Przetwarzanie pliku CSV partiami (chunksize)
Kod poniżej wykonuje operację na danych wariantu 14 (lub na niewielkiej ramce demonstracyjnej, gdy polecenie dotyczy tworzenia ramki).

````python
# Przetwarzanie pliku CSV partiami (chunksize)
liczba_rekordow = 0
suma_mean = 0.0
for fragment in pd.read_csv(plik, chunksize=200):
    liczba_rekordow += len(fragment)
    suma_mean += fragment["mean"].sum()
print("Rekordy:", liczba_rekordow)
print("Średnia mean z partii:", suma_mean / liczba_rekordow)
print("Średnia mean z całej ramki:", df["mean"].mean())
````
